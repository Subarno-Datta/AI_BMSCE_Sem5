#**Tic-Tac-Toe Agent**

```python
def print_board(board):
  print(f"\n {board[0]} | {board[1]} | {board[2]} ")
  print("---|---|---")
  print(f" {board[3]} | {board[4]} | {board[5]} ")
  print("---|---|---")
  print(f" {board[6]} | {board[7]} | {board[8]} \n")
```

```python
def check_win(board, player):
  win_conditions = [
      [0, 1, 2], [3, 4, 5], [6, 7, 8],
      [0, 3, 6], [1, 4, 7], [2, 5, 8],
      [0, 4, 8], [2, 4, 6]
  ]

  for condition in win_conditions:
    if board[condition[0]]==board[condition[1]]==board[condition[2]]==player:
      return True

  return False
```

```python
def play_game():

  board = [str(i) for i in range(1, 10)]
  current_player = 'X'
  turns = 0

  print("Welcome to Tic-Tac-Toe!")
  print_board(board)

  board = [' ' for _ in range(9)]

  while turns < 9:
    try:
      move = int(input(f"Player {current_player}, choose a position (1-9): "))-1

      if move < 0 or move > 8:
        print("invalid input, please choose a number between 1 and 9")
        continue
      elif board[move] != ' ':
        print("That spot is already taken, please try again.")
        continue

      board[move] = current_player
      turns += 1
      print_board(board)

      if check_win(board, current_player):
        print(f"Congratulations! Player {current_player} wins!")
        return

      current_player = 'O' if current_player == 'X' else 'X'

    except ValueError:
      print("Invalid input, please enter a number.")

  print("It's a draw!")
```

```python
if __name__ == "__main__":
  play_game()
```

#**Vacuum Cleaner Agent**

```python
import random
def agent():

  environment_states = ['Dirty', 'Clean']
  environment = {
      'A': random.choice(environment_states),
      'B': random.choice(environment_states)
  }

  current_location = random.choice(['A', 'B'])
  print(f"Initial Environment State: {environment}")
  print(f"Initial Location: {current_location}")

  step = 1

  while True:

    print(f"--- Step {step} ---")
    print(f"Current Location: {current_location}")

    if environment['A'] == 'Clean' and environment['B'] == 'Clean':
        print("Goal achieved: Both rooms are Clean! Shutting down.")
        break

    current_status = environment[current_location]
    print(f"Status of Room {current_location}: {current_status}")

    if current_status == 'Dirty':
      print("Action: SUCK")
      environment[current_location] = 'Clean'

    else:
      if (current_location == 'A'):
        print("Action: MOVE RIGHT to B")
        current_location = 'B'
      elif current_location == 'B':
        print("Action: MOVE LEFT to A")
        current_location = 'A'

    print(f"Environment state after step: {environment}\n")
    step += 1
```

```python
if __name__ == '__main__':
  agent()
```

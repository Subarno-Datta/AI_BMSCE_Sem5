```python
import heapq

class Node:
    def __init__(self, state, parent, action, g, h):
        self.state = state
        self.parent = parent
        self.action = action
        self.g = g
        self.h = h
        self.f = g + h

    def __lt__(self, other):
        return self.f < other.f

def get_neighbors(state):
    neighbors = []
    blank_idx = state.index(0)
    row, col = divmod(blank_idx, 3)

    moves = {'UP': (row - 1, col), 'DOWN': (row + 1, col),
             'LEFT': (row, col - 1), 'RIGHT': (row, col + 1)}

    for action, (r, c) in moves.items():
        if 0 <= r < 3 and 0 <= c < 3:
            new_idx = r * 3 + c
            new_state = list(state)
            new_state[blank_idx], new_state[new_idx] = new_state[new_idx], new_state[blank_idx]
            neighbors.append((tuple(new_state), action))
    return neighbors

def a_star_search_manhattan(start_state, goal_state):
    start_node = Node(start_state, None, None, 0, h_manhattan_distance(start_state, goal_state))
    open_list = []
    heapq.heappush(open_list, start_node)
    closed_set = set()

    while open_list:
        current_node = heapq.heappop(open_list)

        if current_node.state == goal_state:
            path = []
            while current_node.parent:
                path.append(current_node.action)
                current_node = current_node.parent
            return path[::-1]

        closed_set.add(current_node.state)

        for neighbor_state, action in get_neighbors(current_node.state):
            if neighbor_state in closed_set:
                continue

            g_new = current_node.g + 1
            h_new = h_manhattan_distance(neighbor_state, goal_state)
            neighbor_node = Node(neighbor_state, current_node, action, g_new, h_new)

            heapq.heappush(open_list, neighbor_node)

    return None
```

```python
def h_manhattan_distance(state, goal):
    distance = 0
    for i in range(9):
        if state[i] != 0:
          goal_idx = goal.index(state[i])
          curr_row, curr_col = divmod(i, 3)
          goal_row, goal_col = divmod(goal_idx, 3)

          distance += abs(curr_row - goal_row) + abs(curr_col - goal_col)

    return distance
```

```python
def h_misplaced_tiles(state, goal):
    return sum(1 for i in range(9) if state[i] != goal[i] and state[i] != 0)


def a_star_search_misplaced(start_state, goal_state):
    start_node = Node(start_state, None, None, 0, h_misplaced_tiles(start_state, goal_state))
    open_list = []
    heapq.heappush(open_list, start_node)
    closed_set = set()

    while open_list:
        current_node = heapq.heappop(open_list)

        if current_node.state == goal_state:
            path = []
            while current_node.parent:
                path.append(current_node.action)
                current_node = current_node.parent
            return path[::-1]

        closed_set.add(current_node.state)

        for neighbor_state, action in get_neighbors(current_node.state):
            if neighbor_state in closed_set:
                continue

            g_new = current_node.g + 1
            h_new = h_misplaced_tiles(neighbor_state, goal_state)
            neighbor_node = Node(neighbor_state, current_node, action, g_new, h_new)

            heapq.heappush(open_list, neighbor_node)

    return None
```

```python
def run_manhattan():
  initial = (1, 2, 3,
               4, 0, 6,
               7, 5, 8)
  goal = (1, 2, 3,
          4, 5, 6,
          7, 8, 0)

  print("Running A* with Manhattan Distance Heuristic...")
  path = a_star_search_manhattan(initial, goal)
  print(f"Solution found in {len(path)} moves: {path}")
```

```python
def run_misplaced():
  initial = (1, 2, 3,
               4, 0, 6,
               7, 5, 8)
  goal = (1, 2, 3,
          4, 5, 6,
          7, 8, 0)

  print("Running A* with Misplaced Tiles Heuristic...")
  path = a_star_search_misplaced(initial, goal)
  print(f"Solution found in {len(path)} moves: {path}")
```

```python
run_misplaced()
```

```python
run_manhattan()
```

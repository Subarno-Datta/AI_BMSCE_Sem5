```python
def implies(p, q):
    return not p or q

def iff(p, q):
    return (not p or q) and (not q or p)

def tt_entails(kb, alpha, symbols):

    header = " | ".join(symbols) + " || KB | Alpha"
    print(header)
    print("-" * len(header))

    return tt_check_all(kb, alpha, symbols, {}, symbols)

def tt_check_all(kb, alpha, remaining_symbols, model, all_symbols):
    if not remaining_symbols:
        env = {"implies": implies, "iff": iff}

        kb_val = bool(eval(kb, env, model))
        alpha_val = bool(eval(alpha, env, model))

        row_vals = ["T" if model[sym] else "F" for sym in all_symbols]
        row_str = " | ".join(row_vals) + f" || {'T' if kb_val else 'F'}  | {'T' if alpha_val else 'F'}"
        print(row_str)

        if kb_val:
            return alpha_val
        else:
            return True

    else:
        p = remaining_symbols[0]
        rest = remaining_symbols[1:]

        model_true = model.copy()
        model_true[p] = True

        model_false = model.copy()
        model_false[p] = False

        left_branch = tt_check_all(kb, alpha, rest, model_true, all_symbols)
        right_branch = tt_check_all(kb, alpha, rest, model_false, all_symbols)

        return left_branch and right_branch

if __name__ == "__main__":
    print("--- Propositional Logic Truth-Table Enumeration ---\n")

    symbols = ['P', 'Q', 'R']

    kb_expression = "implies(P, Q) and implies(Q, R) and P"
    query_expression = "R"

    print(f"Knowledge Base (KB): {kb_expression}")
    print(f"Query (Alpha): {query_expression}")
    print(f"Symbols: {symbols}\n")

    print("Truth Table:")
    result = tt_entails(kb_expression, query_expression, symbols)

    print("\nConclusion:")
    if result:
        print("Result: TRUE. The Knowledge Base ENTAILS the Query.")
    else:
        print("Result: FALSE. The Knowledge Base DOES NOT entail the Query.")

    print("\n--------------------------------------------------\n")

    kb_expr_2 = "implies(P, Q) and Q"
    query_expr_2 = "P"

    print(f"Knowledge Base (KB): {kb_expr_2}")
    print(f"Query (Alpha): {query_expr_2}")

    print("\nTruth Table:")
    result_2 = tt_entails(kb_expr_2, query_expr_2, ['P', 'Q'])

    print("\nConclusion:")
    print(f"Result: {'TRUE' if result_2 else 'FALSE'}")
```
